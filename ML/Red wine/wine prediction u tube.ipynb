# data preprocessing

```python
#seprate the data and lables
```

```python
X = df.drop('quality',axis=1)
```

```python
print(X)
```

# label binarization

```python
y = df["quality"].apply(lambda y_value: 1 if y_value >= 7 else 0)
```

```python
print(y)
```

# train test split

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(
    X ,y ,test_size = 0.2 ,random_state = 3
)
```

```python
print(y.shape , Y_train.shape , Y_test.shape)
```

# model training Random Forest Classiffier

```python
model = RandomForestClassifier()
```

```python
model.fit(X_train, Y_train)
```

# model evaluation

```python
#accuracy on test data

X_test_prediction = model.predict(X_test)
test_data_accuracy = accuracy_score(X_test_prediction, Y_test)
```

```python
print("accuracy:",test_data_accuracy)
```

# building a predictive system

```python
input_data = (5.4,0.835,0.08,1.2,0.046,13.0,93.0,0.9924,3.57,0.85,13.0)

# changing the input data to a numpy array
input_as_np_array = np.asarray(input_data)

# reshape the data as we are predicting the label for only one instance
input_data_reshaped = input_as_np_array.reshape(1,-1)

prediction = model.predict(input_data_reshaped)
print(prediction)


if (prediction[0]==1):
  print('Good Quality Wine')
else:
  print('Bad Quality Wine')
```
